# 01 · Data Extraction

Downloads historical EPL and La Liga match data from **football-data.co.uk**
and saves the raw combined file to `data/raw/EPL_LaLiga_Data.csv`.

Run this notebook once (or whenever you want to refresh the full dataset).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

from src.data.extract import load_historical_data

RAW_PATH = pathlib.Path("../data/raw/EPL_LaLiga_Data.csv")
```

## Download historical data (2005/06 – 2024/25)

```python
raw = load_historical_data(start_season=2005, end_season=2024)
raw.shape
```

```python
raw.head(3)
```

```python
print("Leagues :", raw["Div"].unique())
print("Seasons :", raw["Season"].nunique())
print("Date range:", raw["Date"].min(), "→", raw["Date"].max())
```

## Save

```python
RAW_PATH.parent.mkdir(parents=True, exist_ok=True)
raw.to_csv(RAW_PATH, index=False)
print(f"Saved {len(raw):,} rows → {RAW_PATH}")
```
